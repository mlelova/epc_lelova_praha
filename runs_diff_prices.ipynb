# Comparing the same scenario to thesis' one

```python
import pypsa

ref = pypsa.Network(
    "solved_networks_core/bat1x_gas22.68_co2_113.4_cy2009_dur1x.nc"
)
```

```python
ref_price = ref.buses_t.marginal_price["DE00"]
ref_slack = ref.generators[
    (ref.generators.bus == "DE00") &
    (ref.generators.marginal_cost >= 2999)
]
fresh = pypsa.Network("small_scale_tests\\baseline_2009_network\\succesful_baseline\\baseline_2030_cy2009_solved.nc")
```

```python
for comp in ["buses", "generators", "storage_units", "links", "loads"]:
    a = getattr(ref, comp)
    b = getattr(fresh, comp)
    print(comp, len(a), len(b))
```

```python
print("ref snapshots:", len(ref.snapshots))
print("fresh snapshots:", len(fresh.snapshots))
```

```python
print("REF total demand TWh:",
      ref.loads_t.p_set.sum().sum() / 1e6)

print("FRESH total demand TWh:",
      fresh.loads_t.p_set.sum().sum() / 1e6)
```

```python
common = ref.generators_t.p_max_pu.columns.intersection(
    fresh.generators_t.p_max_pu.columns
)

diff = (
    ref.generators_t.p_max_pu[common]
    - fresh.generators_t.p_max_pu[common]
).abs()

print("Maximum p_max_pu difference:", diff.max().max())
print("Different values:", (diff > 1e-10).sum().sum())
```

```python
common_storage = ref.storage_units_t.inflow.columns.intersection(
    fresh.storage_units_t.inflow.columns
)

hydro_diff = (
    ref.storage_units_t.inflow[common_storage]
    - fresh.storage_units_t.inflow[common_storage]
).abs()

print("Max hydro inflow difference:", hydro_diff.max().max())
print("Different hydro values:", (hydro_diff > 1e-10).sum().sum())
```

```python
fresh_price = fresh.buses_t.marginal_price["DE00"]

def print_price_stats(label, series):
    print(f"{label}")
    print("Hours:", len(series))
    print("Mean:", series.mean())
    print("Median:", series.median())
    print("Min:", series.min())
    print("Max:", series.max())
    print("Std:", series.std())
    print("-" * 40)

print_price_stats("REF (DE00)", ref_price)
print_price_stats("FRESH (DE00)", fresh_price)
```
